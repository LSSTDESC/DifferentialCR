```python
import numpy as np
import astropy.units as u
import matplotlib.pyplot as plt
import galsim
import os
from astropy.table import Table

from astropy.modeling.models import BlackBody

from utils import BB, apply_filter, apply_DCR, weighted_avg_and_std

from scipy.interpolate import interp1d

def synth_fnu(wl, Flam, band, wl_units = 'nm'):
    '''
    Calculates a synthetic flux
    wl - wavelength array in apropriate units (see wl_units)
    Flam - SED (unfiltered) in in erg/cm^2/s/A
    wl_units - 'nm' or 'A'
    '''

    h  = 6.62607015e-27       # erg s
    c  = 2.99792458e18        # Å / s
    hc = h * c                # erg Å  (≈ 1.986e-8)

    
    if wl_units == 'nm':
        wl_A = wl * 10
        wl_nm = wl
    elif wl_units == 'A':
        wl_A = wl
        wl_nm = wl/10

    filter_file = f'filter_files/total_{band}.dat'
    filter_band = np.loadtxt(filter_file).T #filter_band[0] -> wavelengths, filter_band[1] -> filter pass fraction

    T = interp1d(filter_band[0], filter_band[1])(wl_nm)

    num = np.trapezoid(Flam * T * wl_A, wl_A)
    den = np.trapezoid(T / wl_A, wl_A)

    phot_flux = 1/hc * np.trapezoid(Flam*T*wl_A, wl_A)
    
    return phot_flux # num/ den   # proportional to <f_nu>

h  = 6.62607015e-27       # erg s
c  = 2.99792458e18        # Å / s
hc = h * c                # erg Å  (≈ 1.986e-8)
```

```python

def mean_and_std(x_array, weights_dict, param_array = [None], multi = False):
    if param_array[0] == None:
        param_array = x_array
    
    x_transpose = np.array(param_array).T #create a vertical array for purposes of matrix multiplication below

    means = {}
    stdevs = {}

    if multi:
        for band, weights in weights_dict.items():
            mean = np.trapezoid(x_array, weights * x_transpose, axis = 1)/np.trapezoid(x_array, weights, axis =1)
            means[band] = mean
            stdevs[band] = np.sqrt(np.trapezoid(x_array, weights * (x_transpose - mean[:, np.newaxis])**2, axis=1) / np.trapezoid(x_array, weights, axis=1))

            # print(f'{band} (average): mean = {np.round(np.mean(mean), 4)}, stdev = {np.round(np.mean(stdevs[band]), 4)}')
    else:
        for band, weights in weights_dict.items():
            mean = np.trapezoid(x_array, weights * x_transpose)/np.trapezoid(x_array, weights)
            means[band] = mean
            stdevs[band] = np.sqrt(np.trapezoid(x_array, weights * (x_transpose - mean)**2) / np.trapezoid(x_array, weights))
        

    return means, stdevs
    

flat_refraction_angles, flat_filtered_refracted = apply_DCR(wvl, flat_filtered, zenith, pressure, temperature, H2O_pressure)
BB_refraction_angles, BB_filtered_refracted = apply_DCR(wvl, BB_filtered , zenith, pressure, temperature, H2O_pressure)

flat_means, flat_stdevs = mean_and_std(flat_refraction_angles, flat_filtered_refracted, multi = False)
BB_means, BB_stdevs = mean_and_std(BB_refraction_angles, BB_filtered_refracted, multi = False)

print('Integral over R')
for band in 'g':
    print(band)
    print(BB_means[band])
    print((BB_stdevs[band]**2))


flat_means, flat_stdevs = mean_and_std(wvl, flat_filtered, param_array = flat_refraction_angles, multi = False)
BB_means, BB_stdevs = mean_and_std(wvl, BB_filtered, param_array = BB_refraction_angles, multi = False)

print('Integral over lamdba')
for band in 'g':
    print(band)
    print(BB_means[band])
    print((BB_stdevs[band]**2))

    
# flat_means, flat_stdevs = weighted_avg_and_std(flat_refraction_angles, flat_filtered_refracted, multi = False)
# BB_means, BB_stdevs = weighted_avg_and_std(BB_refraction_angles, BB_filtered_refracted, multi = False)

# print('Sum over R')
# for band in 'g':
#     print(band)
#     print(BB_means[band])
#     print((BB_stdevs[band]**2))


    

plt.plot(BB_refraction_angles, BB_filtered_refracted['g'])
```

```python
# u
# 0.021185486586134242
# g
# 0.026159008485680032
# r
# 0.005596464554058768
# i
# 0.001538581466417939
# z
# 0.00043067589447305253
# y
# 0.0002572913984550017
# define filters and SEDs in galsim

from galsim import degrees
# read in the LSST filters

datapath = galsim.meta_data.share_dir
filter_names = 'ugrizy'
filters = {}
for filter_name in filter_names:
    filter_filename = os.path.join(datapath, 'bandpasses/LSST_{0}.dat'.format(filter_name))
    filter_filename = f'filter_files/total_{filter_name}.dat'
    # Here we create some galsim.Bandpass objects to represent the filters we're observing
    # through.  These include the entire imaging system throughput including the atmosphere,
    # reflective and refractive optics, filters, and the CCD quantum efficiency.  These are
    # also conveniently read in from two-column ASCII files where the first column is
    # wavelength and the second column is dimensionless throughput. The example filter files
    # units of nanometers for the wavelength type, so we specify that using the required
    # `wave_type` argument.
    filters[filter_name] = galsim.Bandpass(filter_filename, wave_type='nm')
    # For speed, we can thin out the wavelength sampling of the filter a bit.
    # In the following line, `rel_err` specifies the relative error when integrating over just
    # the filter (however, this is not necessarily the relative error when integrating over the
    # filter times an SED).
    filters[filter_name] = filters[filter_name].thin(rel_err=1e-4)
    # plt.plot(filters[filter_name].wave_list, filters[filter_name].__call__(filters[filter_name].wave_list))


table = galsim.LookupTable(wvl*10, BB5000, interpolant='linear')
sed = galsim.SED(table, wave_type='Ang', flux_type='flambda')


table_flat = galsim.LookupTable(wvl*10, flat, interpolant='linear')
sed_flat = galsim.SED(table_flat, wave_type='Ang', flux_type='flambda')

plt.plot(np.linspace(320, 1000, 1000), sed._mul_bandpass(filters['g'])._call_fast(np.linspace(320, 1000, 1000))/np.max(sed._mul_bandpass(filters['g'])._call_fast(np.linspace(320, 1000, 1000))))
plt.plot(wvl, BB_filtered['g']/np.max(BB_filtered['g']), ls = '-', color = colors['g'])

    
```
